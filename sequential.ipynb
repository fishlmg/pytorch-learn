```python
import torch
import torch.nn as nn
```

```python
# 定义模型
class AreaModel(nn.Module):
    def __init__(self):
        super(AreaModel, self).__init__()
        self.model = nn.Sequential(
            nn.Linear(2, 128),
            nn.ReLU(),
            nn.Linear(128, 256),
            nn.ReLU(),
            nn.Linear(256, 128),
            nn.ReLU(),
            nn.Linear(128, 1)
        )

    def forward(self, x):
        return self.model(x)

model = AreaModel()
print(model)
```
